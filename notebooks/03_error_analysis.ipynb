# 03. error_analysis
**Project**: Recut Research Lab 01  
**Status**: Research Notebook Skeleton (Protocol Design)  

---

### 1. Objective
Decompose baseline prediction errors on the validation split into structural failure categories (e.g. lexical overlap illusion, paraphrase gap, temporal/numeric mismatch).

---

### 2. Research Questions
- RQ: What proportion of baseline errors arise from shallow lexical overlap vs. fine-grained claim discrepancies?
- RQ: Do false positives cluster around specific text length quantiles or high-confidence prediction thresholds?

---

### 3. Hypotheses
- H1: The majority of False Positives exhibit high lexical overlap, failing because the LP mentions the keyword without supporting the headline predicate.
- H2: False Negatives correlate with creative paraphrase expressions absent from the raw LP.

---

### 4. Methodology
1. Collect misclassified validation instances from baseline models.
2. Stratify errors by prediction confidence and overlap features.
3. Categorize errors against the standard error taxonomy (E1-E6).
4. Tabulate category distribution and identify mitigation directions.

---

### 5. Required Inputs
- Trained baseline model checkpoint (`models/tfidf_logreg_baseline.pkl`)
- Validation split DataFrame (`data/raw/adtec_consistency/valid.jsonl`)
- Prediction error records (`experiments/runs/*/predictions.csv`)

---

### 6. Expected Outputs
- Error distribution table across failure categories E1-E6.
- Confusion slice charts (error rate vs. text length, error rate vs. lexical overlap).
- Qualitative case audit log.

---

### 7. Limitations & Scientific Guardrails
Error analysis must be performed strictly on the validation set. Never expose or analyze errors on the test split prior to model freezing.
