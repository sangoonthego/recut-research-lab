**Project**: Recut Research Lab 01  
**Status**: Research Notebook Skeleton (Protocol Design)  

---

### 1. Objective
Evaluate the out-of-domain transferability of AdTEC-trained consistency models to generated ad copy in Recut product verticals.

---

### 2. Research Questions
- RQ: How severe is the performance degradation when evaluating AdTEC models on generated promotional copy from new vertical domains?
- RQ: Can zero-shot lexical constraints or lightweight adapter tuning mitigate domain shift?

---

### 3. Hypotheses
- H1: Models reliant on search ad keywords will suffer precision drops when exposed to marketing copy with emotive or lifestyle appeals.

---

### 4. Methodology
1. Curate a gold-standard Recut evaluation set (`data/recut_eval/`).
2. Evaluate frozen baseline and enhanced models zero-shot.
3. Quantify domain transfer gap across macro-F1 and per-class precision.

---

### 5. Required Inputs
- AdTEC trained models.
- Recut evaluation split (`data/recut_eval/recut_test.jsonl`).

---

### 6. Expected Outputs
- In-domain vs. Out-of-domain performance comparison matrix.
- Transfer gap diagnostics.

---
